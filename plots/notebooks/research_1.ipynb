```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.ticker import MaxNLocator

plt.rcParams.update({
    "font.family": "serif",
    "font.size": 9,
    "axes.labelsize": 9,
    "axes.titlesize": 9,
    "xtick.labelsize": 8,
    "ytick.labelsize": 8,
    "legend.fontsize": 8,
    "axes.linewidth": 0.8,
    "xtick.major.width": 0.8,
    "ytick.major.width": 0.8,
})

ARCH_ORDER = [
    "Direct LUT",
    "LUT + Interpolation",
    "Newton-Raphson",
    "Goldschmidt",
    "Hybrid",
]

FMT_ORDER = ["Q8.24", "Q16.16"]

df = df.copy()

df["Algorithm"] = pd.Categorical(
    df["Algorithm"],
    categories=ARCH_ORDER,
    ordered=True
)

df["Configuration"] = pd.Categorical(
    df["Configuration"],
    categories=FMT_ORDER,
    ordered=True
)

df = df.sort_values(["Algorithm", "Configuration"]).reset_index(drop=True)

# Clean known numeric fields if they arrived as strings.
numeric_cols = [
    "Int bits", "Frac bits", "Clock Target (ns)",
    "Seed LUT size (bits)", "Target Freq (MHz)",
    "Max Abs Error", "MAE", "RMSE (LSB)",
    "Max Error (LSB)", "Max Relative Error (%)",
    "LUT", "LUT Usage (%)", "FF", "DSP", "BRAM",
    "Fmax (MHz)", "WNS", "Latency (cycles)",
    "Latency (ns)", "Max Throughput (Mops/s)",
    "Dynamic Power (W)", "Static Power (W)",
    "Total Power (W)", "Energy/op (nJ)"
]

for col in numeric_cols:
    if col in df.columns:
        df[col] = pd.to_numeric(
            df[col].astype(str).str.replace(",", "", regex=False)
            .str.replace(" LSB", "", regex=False)
            .str.rstrip("."),
            errors="coerce"
        )

df
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams.update({
    "font.family": "serif",
    "font.serif": ["Times New Roman", "Times", "DejaVu Serif"],
    "mathtext.fontset": "stix",
    "font.size": 8.5,
    "axes.labelsize": 9,
    "axes.titlesize": 9,
    "xtick.labelsize": 8,
    "ytick.labelsize": 8,
    "legend.fontsize": 7.5,
    "axes.linewidth": 0.75,
    "xtick.major.width": 0.65,
    "ytick.major.width": 0.65,
    "xtick.major.size": 3,
    "ytick.major.size": 3,
    "figure.dpi": 150,
    "savefig.dpi": 600
})

ARCH_ORDER = [
    "Direct LUT",
    "LUT + Interpolation",
    "Newton-Raphson",
    "Goldschmidt",
    "Hybrid"
]

ARCH_SHORT = {
    "Direct LUT": "LUT",
    "LUT + Interpolation": "INT",
    "Newton-Raphson": "NR",
    "Goldschmidt": "GS",
    "Hybrid": "HYB"
}

COLORS = {
    "Q8.24": "#1F4E79",
    "Q16.16": "#8C2D2D"
}

MARKERS = {
    "Q8.24": "o",
    "Q16.16": "s"
}

FMT_LABEL = {
    "Q8.24": r"$Q_{8.24}$",
    "Q16.16": r"$Q_{16.16}$"
}

plot_df = df.copy()

numeric_cols = [
    "LUT", "FF", "DSP", "BRAM",
    "Fmax (MHz)", "WNS",
    "Latency (cycles)", "Latency (ns)",
    "Max Abs Error", "MAE",
    "RMSE (LSB)", "Max Error (LSB)",
    "Max Relative Error (%)",
    "Total Power (W)", "Energy/op (nJ)",
    "Max Throughput (Mops/s)"
]

for col in numeric_cols:
    plot_df[col] = pd.to_numeric(
        plot_df[col].astype(str)
        .str.replace(",", "", regex=False)
        .str.replace("%", "", regex=False)
        .str.rstrip("."),
        errors="coerce"
    )

plot_df["Architecture"] = plot_df["Algorithm"].map(ARCH_SHORT)

def clean(ax):
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)
    ax.grid(axis="y", alpha=0.18, linewidth=0.5)
    ax.set_axisbelow(True)
```

```python
fig, axes = plt.subplots(
    2, 2,
    figsize=(7.16, 4.8),
    constrained_layout=True
)

resources = [
    ("LUT", "LUT count"),
    ("FF", "Flip-flop count"),
    ("DSP", "DSP count"),
    ("BRAM", "BRAM count")
]

x = np.arange(len(ARCH_ORDER))

for ax, (column, ylabel) in zip(axes.flat, resources):

    for fmt in ["Q8.24", "Q16.16"]:

        sub = (
            plot_df[
                plot_df["Configuration"] == fmt
            ]
            .set_index("Algorithm")
            .reindex(ARCH_ORDER)
        )

        ax.plot(
            x,
            sub[column],
            color=COLORS[fmt],
            marker=MARKERS[fmt],
            markersize=4.5,
            linewidth=1.4,
            markerfacecolor="white",
            markeredgewidth=0.9,
            label=FMT_LABEL[fmt]
        )

    ax.set_xticks(x)
    ax.set_xticklabels(
        [ARCH_SHORT[a] for a in ARCH_ORDER]
    )

    ax.set_ylabel(ylabel)
    clean(ax)

handles, labels = axes[0, 0].get_legend_handles_labels()

fig.legend(
    handles,
    labels,
    loc="upper center",
    bbox_to_anchor=(0.5, 1.04),
    ncol=2,
    frameon=False
)

plt.savefig("/content/resources_plot.png", bbox_inches="tight")
plt.show()
```

```python
pareto_df = plot_df.copy()

metrics = {
    "Accuracy": ("Max Relative Error (%)", False),
    "Latency": ("Latency (ns)", False),
    "LUT": ("LUT", False),
    "Energy": ("Energy/op (nJ)", False)
}

for name, (column, higher_better) in metrics.items():

    values = pareto_df[column]

    if higher_better:
        pareto_df[name] = (
            values - values.min()
        ) / (
            values.max() - values.min()
        )
    else:
        pareto_df[name] = (
            values.max() - values
        ) / (
            values.max() - values.min()
        )

pareto_df["Composite"] = pareto_df[
    ["Accuracy", "Latency", "LUT", "Energy"]
].mean(axis=1)
fig, ax = plt.subplots(
    figsize=(7.16, 3.3)
)

x = np.arange(len(pareto_df))

for fmt in ["Q8.24", "Q16.16"]:

    sub = pareto_df[
        pareto_df["Configuration"] == fmt
    ]

    indices = [
        pareto_df.index.get_loc(i)
        for i in sub.index
    ]

    ax.plot(
        indices,
        sub["Composite"],
        color=COLORS[fmt],
        marker=MARKERS[fmt],
        markersize=5,
        linewidth=1.5,
        markerfacecolor="white",
        markeredgewidth=0.9,
        label=FMT_LABEL[fmt]
    )

ax.set_xticks(x)
ax.set_xticklabels(
    pareto_df["Architecture"]
)

ax.set_ylabel("Normalized design-space score")

clean(ax)

ax.legend(
    frameon=False,
    loc="best"
)

plt.tight_layout()
plt.savefig("/content/pareto_score_plot.png", bbox_inches="tight")
plt.show()
```

```python
from math import pi
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Define ARCH_LS, as it's used in this cell but not defined previously
ARCH_LS = {
    "Direct LUT": "-",
    "LUT + Interpolation": "--",
    "Newton-Raphson": "-.",
    "Goldschmidt": ":",
    "Hybrid": (0, (3, 5, 1, 5, 1, 5)),
}

# Define the metrics for radar chart labels and their corresponding plot_df columns
# along with whether a higher value is better (for normalization)
radar_normalization_metrics = {
    r"$F_{\mathrm{max}}$": ("Fmax (MHz)", True),
    r"$T_{\mathrm{lat}}$": ("Latency (ns)", False),
    r"$N_{\mathrm{LUT}}$": ("LUT", False),
    r"$N_{\mathrm{DSP}}$": ("DSP", False),
    r"$P_{\mathrm{tot}}$": ("Total Power (W)", False),
    r"$E_{\mathrm{op}}$": ("Energy/op (nJ)", False),
    "Error": ("Max Relative Error (%)", False) # Using Max Relative Error as the 'Error' metric
}

rada_df = plot_df.copy() # Using plot_df as the base

# Normalize the data for radar chart
for label, (column_name, higher_is_better) in radar_normalization_metrics.items():
    min_val = rada_df[column_name].min()
    max_val = rada_df[column_name].max()

    if (max_val - min_val) == 0:
        rada_df[label] = 0.5 # Assign a neutral value if all values are the same
    elif higher_is_better:
        rada_df[label] = (rada_df[column_name] - min_val) / (max_val - min_val)
    else:
        rada_df[label] = (max_val - rada_df[column_name]) / (max_val - min_val)

radar_metrics = list(radar_normalization_metrics.keys())

N = len(radar_metrics)

angles = [
    n / float(N) * 2 * pi
    for n in range(N)
]

angles += angles[:1]

fig, axes = plt.subplots(
    1,
    2,
    figsize=(7.16, 3.5),
    subplot_kw=dict(polar=True)
)

for ax, fmt in zip(
    axes,
    ["Q8.24", "Q16.16"]
):

    # Changed heat_df to rada_df and use the normalized dataframe
    sub = rada_df[
        rada_df["Configuration"] == fmt
    ]

    for _, row in sub.iterrows():

        # Extract normalized values directly from rada_df using radar_metrics labels
        values = [row[metric_label] for metric_label in radar_metrics]
        values += values[:1]

        ax.plot(
            angles,
            values,
            linewidth=1.0,
            color="black",
            linestyle=ARCH_LS.get(
                row["Algorithm"],
                "-"
            ),
            marker="o",
            markersize=2.5,
            markerfacecolor="white",
            markeredgecolor="black"
        )

    ax.set_xticks(
        angles[:-1]
    )

    ax.set_xticklabels(
        radar_metrics,
        fontsize=6.5
    )

    ax.set_ylim(0, 1)

    ax.set_yticks(
        [0.25, 0.5, 0.75, 1.0]
    )

    ax.set_yticklabels(
        ["0.25", "0.50", "0.75", "1.00"],
        fontsize=5.5
    )

    ax.set_title(
        FMT_LABEL[fmt],
        pad=12
    )

fig.legend(
    [
        plt.Line2D(
            [0],
            [0],
            color="black",
            linestyle=ARCH_LS[a],
            linewidth=1.0,
            marker="o",
            markersize=2.5,
            markerfacecolor="white"
        )
        for a in ARCH_ORDER
    ],
    [
        ARCH_SHORT[a]
        for a in ARCH_ORDER
    ],
    loc="lower center",
    bbox_to_anchor=(0.5, -0.01),
    ncol=5,
    frameon=False
)

plt.subplots_adjust(
    bottom=0.17,
    top=0.88,
    wspace=0.2
)

plt.savefig("/content/radar_chart_plot.png", bbox_inches="tight")
plt.show()
```
